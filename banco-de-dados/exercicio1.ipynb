```python
import pandas as pd

df = pd.read_csv('../data-sets/2025-09-01_Alvaras_-_Base_de_Dados.csv', sep=';', encoding='latin1')

df.head()
```

```python
for col in df.columns:
    if df[col].nunique() == len(df):
        print(f"{col} could be a primary key (all unique values)")
```

```python
# Concatanate NOME_EMPRESARIAL + NUMERO_DO_ALVARA
df['potential_key'] = df['NOME_EMPRESARIAL'] + df['NUMERO_DO_ALVARA'].astype(str)

if df['potential_key'].nunique() == len(df):
    print("Combination of NOME_EMPRESARIAL and NUMERO_DO_ALVARA could be a primary key (all unique values)")
```
